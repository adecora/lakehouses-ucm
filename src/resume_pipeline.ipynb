```python
catalog = dbutils.widgets.get("catalog")
schema_prefix = dbutils.widgets.get("schema_prefix")
```

```python
tables_list = spark.sql(f"""
                    SELECT table_schema || '.' || table_name as table_name, table_type
                    FROM system.information_schema.tables
                    WHERE table_catalog = '{catalog}'
                    AND table_schema LIKE '{schema_prefix}'
                    AND table_type IN ('STREAMING_TABLE', 'MATERIALIZED_VIEW')
                """).collect()
```

```python
query = [f"SELECT '{t.table_name}' as table_name, COUNT(*) as row_count FROM {t.table_name}" for t in tables_list]
spark.sql(" UNION ALL ".join(query)).show(truncate=False)
```
